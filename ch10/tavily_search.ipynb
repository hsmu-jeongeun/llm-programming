## TAVILY 검색 도구 설정

```python
from langchain_tavily import TavilySearch

tavily_search = TavilySearch(
    max_results=5 # 검색 결과 몇개 가져올지 설정
)
```

## 검색 및 출력

```python
res = tavily_search.invoke({"query": "2027 한국 경제 전망"})
for r in res["results"]:
    print(r)
```

- 가독성 좋게 출력

```python
from IPython.display import JSON
JSON(res).data
```

## 웹페이지 전문 불러오기

```python
from tavily import TavilyClient

client = TavilyClient()

content = client.search(
    "2027년 한국 경제 전망",  
    search_depth="advanced",
    include_raw_content=True
)

JSON(content).data
```

## 질문에 대한 자료 조사하기
- 질문에 대한 자료를 TavilyClient로 가져오기

```python
from langchain_core.messages import convert_to_messages
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-4o"
                 , default_headers={"Accept-Encoding": "identity"})

query = "현대자동차 미국 시장 2027년 전망"

content = client.search(
    query, 
    include_raw_content=True,
    search_depth="advanced",
    max_results=1 
)["results"]

content
```

## 프롬프트 설정

```python
prompt = [{
    "role": "system",
    "content":  (
        "당신은 신문기사를 쓰는 기자 AI입니다.  \n"
        "당신은 주어진 정보를 바탕으로 객관적이고 체계적으로 작성된 기사를 써야 합니다. \n"
    )
}, {
    "role": "user",
    "content": (
        f'정보: """{content}"""\n\n' 
        f'위의 정보를 사용하여, 다음 질문에 대해 자세한 보고서를 한국어로 작성하세요: "{query}" \n'
        '—신문기사 형식을 사용하되, MLA를 준수하는 markdown 문법을 사용해주세요.'
        '—활용한 자료는 출처를 명시하세요.'
    )
    
}]
```
